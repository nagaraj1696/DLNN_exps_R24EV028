```python
import numpy as np

# Step activation function
def step_function(x):
    return 1 if x >= 0 else 0


# Single-Layer Perceptron
def perceptron(X, y, learning_rate=0.1, epochs=10):

    # Initialize weights and bias
    weights = np.zeros(X.shape[1])
    bias = 0

    for epoch in range(epochs):
        total_error = 0

        for i in range(len(X)):
            # Calculate weighted sum
            z = np.dot(X[i], weights) + bias

            # Apply step activation
            y_pred = step_function(z)

            # Calculate error
            error = y[i] - y_pred
            total_error += abs(error)

            # Update weights and bias
            weights = weights + learning_rate * error * X[i]
            bias = bias + learning_rate * error

        print(f"Epoch {epoch + 1}: Error = {total_error}")

        # Stop if no error
        if total_error == 0:
            break

    return weights, bias


# Function to test the perceptron
def test_perceptron(X, weights, bias):
    predictions = []

    for sample in X:
        z = np.dot(sample, weights) + bias
        predictions.append(step_function(z))

    return predictions


# -------------------------------------------------
# AND Gate
# -------------------------------------------------
X_and = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_and = np.array([0, 0, 0, 1])

weights, bias = perceptron(X_and, y_and)

print("\nAND Gate")
print("Weights:", weights)
print("Bias:", bias)
print("Output:", test_perceptron(X_and, weights, bias))


# -------------------------------------------------
# OR Gate
# -------------------------------------------------
X_or = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_or = np.array([0, 1, 1, 1])

weights, bias = perceptron(X_or, y_or)

print("\nOR Gate")
print("Weights:", weights)
print("Bias:", bias)
print("Output:", test_perceptron(X_or, weights, bias))


# -------------------------------------------------
# NOT Gate
# -------------------------------------------------
X_not = np.array([
    [0],
    [1]
])

y_not = np.array([1, 0])

weights, bias = perceptron(X_not, y_not)

print("\nNOT Gate")
print("Weights:", weights)
print("Bias:", bias)
print("Output:", test_perceptron(X_not, weights, bias))
```
